# Handling missing value

```python
def missing_values(data : pd.DataFrame, threshold : float = 10):

    na_count = data.isna().sum()
    na_exist = na_count[na_count > 0]
    na_abs_frq = na_exist.values
    na_rel_frq = round(na_exist/len(data)*100)
    missings = pd.DataFrame({'Nº of missings': na_abs_frq, '% of missings': na_rel_frq})
    missings = missings.sort_values(by = 'Nº of missings', ascending = False)

    if threshold:
        missings['Above threshold'] = [True if x > threshold else False for x in missings['% of missings']]
    
    return missings

missing_values(df_cleaned)
```

## using Spotify API

```python
# 1. define columns we want to fill
fill_columns = [
    "lyrics", "featured_artists", "album", "language", 
    "year", "month", "day", "bpm", "rolloff", "rms", "flux",
    "spectral_complexity", "pitch", "flatness", "loudness", "album_name",
    "album_type", "disc_number", "track_number", "album_release_date",
    "id_album", "duration_ms", "explicit", "popularity"
]

# 2. Remove duplicates in spotify API based on id + id_artist
spotify_clean = (
    spotifyAPI
    .sort_values(by=fill_columns, ascending=False)
    .drop_duplicates(subset=["id", "id_artist"], keep="first")
)

# 3. Build lookup dictionary
spotify_lookup = spotify_clean.set_index(["id", "id_artist"]).to_dict(orient="index")

# 4. Fill missing values in df_cleaned
filled_rows = 0

for idx, row in df_cleaned.iterrows():
    key = (row["id"], row["id_artist"])

    if key not in spotify_lookup:
        continue

    source = spotify_lookup[key]
    updated = False

    for col in fill_columns:
        val = row[col]

        is_missing = (
            (isinstance(val, float) and pd.isna(val))
            or val is None
            or val == ""
            or val == []
        )

        if is_missing and col in source and not pd.isna(source[col]):
            df_cleaned.at[idx, col] = source[col]
            updated = True

    if updated:
        filled_rows += 1

print(f"Filled missing values for {filled_rows} rows using spotifyAPI.")
```

- Infer from title or album_name or album (if they have text like 'feat.')

```python
import re
def normalize_name(s):
    """Normalize single artist name:
       - remove surrounding parentheses/brackets
       - remove leading non-alphanumeric punctuation (like ". " or " -")
       - collapse whitespace
       - sensible title-casing
    """
    if not isinstance(s, str):
        return s
    s = s.strip()
    # remove surrounding parentheses/brackets
    s = s.strip("()[]{} \u2018\u2019\u201c\u201d")
    # remove leading non-alphanumeric characters (dots, punctuation, etc.)
    s = re.sub(r'^[^0-9A-Za-zÀ-ÖØ-öø-ÿ]+', '', s)
    # collapse multiple spaces
    s = re.sub(r'\s+', ' ', s)
    # Remove trailing punctuation like "." or ","
    s = re.sub(r'[\.\,\;\:\-\/\|]+$','', s).strip()
    # Title case (keeps accents)
    return s.title()

def split_artists(raw):
    """
    Split a raw extracted chunk into individual artist names using common separators.
    Returns comma-separated string of cleaned names, or None if nothing valid.
    """
    if not raw or not isinstance(raw, str):
        return None
    # split on common separators (longer tokens first)
    parts = re.split(r'\s*(?:,|&|;|\/| and | feat | ft | featuring )\s*', raw, flags=re.IGNORECASE)
    names = []
    for p in parts:
        p = p.strip()
        if not p:
            continue
        # drop surrounding parentheses fragments
        p = re.sub(r'^\(+|\)+$','', p).strip()
        # remove leading punctuation (fixes ". Shiva" case)
        p = re.sub(r'^[^0-9A-Za-zÀ-ÖØ-öø-ÿ]+', '', p)
        # remove trailing descriptors like "feat", "ft" leftover
        p = re.sub(r'\b(?:feat|ft|featuring)\b[\.]?$','', p, flags=re.IGNORECASE).strip()
        if p:
            names.append(normalize_name(p))
    if not names:
        return None
    # deduplicate preserving order
    seen = set()
    out = []
    for n in names:
        if n not in seen:
            seen.add(n)
            out.append(n)
    return ", ".join(out) if out else None

def extract_feat(text):
    """
    Extract featured artists from text:
    - requires whole-word feat/ft/featuring (word boundaries)
    - captures artist chunk up to ) , ; - : or end
    Returns cleaned comma-separated string or None.
    """
    if not isinstance(text, str):
        return None
    txt = text.strip()

    # Match tokens like "feat.", "ft.", "featuring" as separate words and capture following chunk.
    pattern = re.compile(
        r"\b(?:feat(?:\.|uring)?|ft(?:\.)?)\b[:\s\-–—]*([^)\;\,\-:]{1,300})",
        flags=re.IGNORECASE
    )
    matches = pattern.findall(txt)
    if not matches:
        # fallback: sometimes featured artists are inside parentheses without the word "feat"
        # e.g. "EVERYDAY (Shiva, ANNA, Geolier)" -> capture inside parentheses if comma separated and length small
        paren_match = re.search(r"\(([^)]+)\)", txt)
        if paren_match:
            # only use if likely contains multiple artists or common separators
            candidate = paren_match.group(1)
            if re.search(r"\b(?:,|&|ft|feat|featuring|and|/)\b", candidate, flags=re.IGNORECASE):
                return split_artists(candidate)
        return None

    combined = " / ".join(m.strip() for m in matches if m.strip())
    return split_artists(combined)

def fill_featured_artists(df):
    """
    Fill missing 'featured_artists' by extracting from title/album_name/album.
    Logs each fill as: Row idx | 'source_text' → 'Artist1, Artist2' (from source_column)
    """
    df = df.copy()
    text_fields = ["title", "album_name", "album"]
    filled = []

    for idx, row in df.iterrows():
        current = row.get("featured_artists")
        if pd.notna(current) and str(current).strip() != "":
            continue

        extracted = None
        used_col = None
        used_text = None

        for col in text_fields:
            src_text = row.get(col)
            if not isinstance(src_text, str):
                continue
            feat = extract_feat(src_text)
            if feat:
                extracted = feat
                used_col = col
                used_text = src_text
                break

        if extracted:
            df.at[idx, "featured_artists"] = extracted
            # prepare snippet for logging (truncate if very long)
            snip = (used_text[:140] + "...") if used_text and len(used_text) > 140 else (used_text or "")
            filled.append((idx, snip, extracted, used_col))

    # Print detailed log using the actual source column content
    print("\nFEATURED ARTISTS INFERRED FROM TEXT:\n")
    for idx, src_text, artist, source in filled:
        print(f"Row {idx} | '{src_text}' → '{artist}' (from {source})")
    print(f"\nTotal inferred rows: {len(filled)}")

    return df
```

```python
df_cleaned = fill_featured_artists(df_cleaned)
```

## n_tokens, n_sentences ,... 

- Compute from 'lyrics'

```python
def compute_text_features(df):
    """
    Compute and fill NA for:
        - n_tokens
        - n_sentences
        - char_per_tok
        - avg_token_per_clause
    based on the 'lyrics' column.
    """
    df = df.copy()
    
    # Prepare empty columns
    df["n_tokens"] = 0
    df["n_sentences"] = 0
    df["char_per_tok"] = 0.0
    df["avg_token_per_clause"] = 0.0

    for idx, text in df["lyrics"].items():

        # Handle missing values
        if not isinstance(text, str) or text.strip() == "":
            continue

        # -------------------------------
        # 1) TOKENS (simple whitespace split)
        # -------------------------------
        tokens = re.findall(r"\b\w+\b", text, flags=re.UNICODE)
        n_tokens = len(tokens)

        # -------------------------------
        # 2) SENTENCES (split by ., ?, !)
        # -------------------------------
        sentences = re.split(r"[.!?]+", text)
        sentences = [s.strip() for s in sentences if s.strip() != ""]
        n_sentences = len(sentences)

        # -------------------------------
        # 3) avg character length per token
        # -------------------------------
        if n_tokens > 0:
            char_per_tok = sum(len(t) for t in tokens) / n_tokens
        else:
            char_per_tok = 0.0

        # -------------------------------
        # 4) avg tokens per clause
        #    Clauses = split by , ; : – or newline
        # -------------------------------
        clauses = re.split(r"[,;:\-\n]+", text)
        clauses = [c.strip() for c in clauses if c.strip() != ""]
        n_clauses = len(clauses)

        if n_clauses > 0:
            avg_token_per_clause = n_tokens / n_clauses
        else:
            avg_token_per_clause = 0.0

        # ---- assign to df ----
        df.at[idx, "n_tokens"] = n_tokens
        df.at[idx, "n_sentences"] = n_sentences
        df.at[idx, "char_per_tok"] = round(char_per_tok, 4)
        df.at[idx, "avg_token_per_clause"] = round(avg_token_per_clause, 4)

    return df
```

## language

- Detect language directly from "lyrics"
- If multiple languages appear → choose the language with the highest percentage

```python
from langdetect import detect_langs

def detect_main_language(text):
    """
    Detect main language from lyrics:
    - Uses langdetect → returns multiple languages with probabilities
    - Returns the language with highest probability
    - Fallback: simple heuristic on character frequency
    """
    if not isinstance(text, str) or text.strip() == "":
        return "unknown"

    # Clean text
    cleaned = re.sub(r"[^A-Za-zÀ-ÖØ-öø-ÿ ]+", " ", text)

    # Try langdetect
    try:
        langs = detect_langs(cleaned)
        if len(langs) > 0:
            # Example: [en:0.73, it:0.27]
            langs_sorted = sorted(langs, key=lambda x: x.prob, reverse=True)
            return langs_sorted[0].lang
    except:
        pass

    # ------------------------
    # Fallback heuristic
    # ------------------------
    text_low = cleaned.lower()

    italian_chars = sum(text_low.count(c) for c in "àèéìòóù")
    english_chars = sum(text_low.count(c) for c in "wkjqxz")  # rare in Italian

    if italian_chars > english_chars:
        return "it"
    elif english_chars > italian_chars:
        return "en"
    else:
        return "unknown"


def fill_language_column(df):
    df = df.copy()
    
    df["language"] = df["lyrics"].apply(detect_main_language)

    return df
```

```python
def classify_title(title):
    if pd.isna(title):
        return "missing"
    
    t = str(title).strip()

    # 1. Hashtag
    if "#" in t:
        return "hashtag"

    # 2. Emoji-heavy
    if re.search(r"[\U0001F600-\U0001F64F\u2600-\u27BF]", t):
        return "emoji"

    # 3. Freestyle
    if "freestyle" in t.lower():
        return "freestyle"

    # 4. Red Bull freestyle
    if "64 barre" in t.lower():
        return "redbull_freestyle"

    # 5. Mashup
    if "mashup" in t.lower():
        return "mashup"

    # 6. Versioning: Live, Demo, Opera, Jam
    if re.search(r"\((live|demo|opera|jam|unplugged|version)", t.lower()):
        return "performance_version"

    # 7. Date-like patterns
    if re.match(r"^\d{1,2}[\/\-]\d{1,2}[\/\-]\d{2,4}$", t):
        return "date"

    # 8. Numeric-only
    if re.match(r"^\d+$", t):
        return "numeric"

    # 9. Math / formula
    if re.search(r"[=+\-\^\(\)\/\*]", t) and any(c.isdigit() for c in t):
        return "formula/math"

    # 10. Symbols only
    if re.match(r"^[\+\-\=\.\*\/\^]+$", t):
        return "symbols"

    return "standard"


def classify_titles(df):
    df = df.copy()
    df["title_type"] = df["title"].apply(classify_title)
    return df
```

```python
pattern_artist = r"^ART\d+$"
pattern_album  = r"^ALB\d+$"
pattern_track  = r"^TR\d+$"

def check_regex(df, col, pattern):
    invalid = df[~df[col].astype(str).str.fullmatch(pattern)]
    print(f"Invalid format for {col}: {len(invalid)} rows")
    return invalid[[col]]
```

```python
def fix_track_id(id_value):
    if not isinstance(id_value, str):
        return id_value
    
    m = re.match(r"^TR_(\d+)$", id_value.strip(), re.IGNORECASE)
    if m:
        return "TR" + m.group(1)
    
    return id_value

df_cleaned["id"] = df_cleaned["id"].apply(fix_track_id)
```

```python
def fix_TR367132(df):
    """
    Split TR367132 into 2 tracks exactly according to album:
    - BUGIE -> album_type contains 'album' (keep id)
    - sentimi -> album_type contains 'single' (change id)
    """
    
    # ---- HÀM CLEAN TITLE ----
    def clean_title(s):
        if pd.isna(s):
            return s
        s = str(s)
        for zw in ["\u200b", "\u200c", "\u200d", "\ufeff"]:  # ZERO-WIDTH + BOM
            s = s.replace(zw, "")
        return s.strip().lower()

    sub = df[df["id"] == "TR367132"].copy()
    if sub.empty:
        return df

    # Tạo cột đã normalize
    sub["title_clean"] = sub["title"].apply(clean_title)
    
    # ======================
    #   TÌM BUGIE
    # ======================
    bugie_row = sub[
        (sub["title_clean"] == "bugie") &
        (sub["album_type"].str.lower().str.contains("album", na=False))
    ]

    bugie_fixed = bugie_row.iloc[0].copy()
    bugie_fixed["id"] = "TR367132"

    # ======================
    #   TÌM SENTIMI
    # ======================
    sentimi_row = sub[
        (sub["title_clean"] == "sentimi") &
        (sub["album_type"].str.lower().str.contains("single", na=False))
    ]

    sentimi_fixed = sentimi_row.iloc[0].copy()
    sentimi_fixed["id"] = "TR999927"

    # Build final output
    fixed = pd.DataFrame([bugie_fixed, sentimi_fixed])

    df_rest = df[df["id"] != "TR367132"].copy()
    df_out = pd.concat([df_rest, fixed], ignore_index=True)

    return df_out[df.columns]
```

```python
def reassign_duplicate_ids(df, start_new_id="TR999928"):
    """
    For each id that appears multiple times:
    - keep the first line
    - assign new, increasing id to the remaining duplicate lines: TR999928, TR999929...
    """
    df = df.copy()

    prefix = start_new_id[:2]   # TR
    counter = int(start_new_id[2:])  # 999928

    new_ids = []

    for id_val, group in df.groupby("id"):
        if len(group) == 1:
            continue  

        keep_idx = group.index[0]

        dup_idx = group.index[1:]

        for idx in dup_idx:
            new_id = f"{prefix}{counter}"
            counter += 1
            new_ids.append((idx, new_id))

    for idx, new_id in new_ids:
        df.at[idx, "id"] = new_id

    return df
```

- Same 'id_artist', same 'album' (name), same 'album_release_date' => same 'id_album'

1. If id_album is in the correct format (ALB123456) → keep it.
2. If the format is incorrect → set to NA. 
3. Group by (id_artist, album, album_release_date).
4. If the group has:
    - 1 valid id_album → use for the whole group.
    - more than 1 id_album → take the most frequently appearing ID (mode).
    - 0 id_album → create a new id_album: ALB999843, ALB999844, ALB999845, ...

```python
# 1. Clean and extract the numeric part: This solves the NameError and potential ValueErrors.
numeric_ids = df_cleaned['id_album'].str.replace('ALB', '', regex=False).str.strip()
df_cleaned['id_numeric'] = pd.to_numeric(numeric_ids, errors='coerce')

# 2. Find the maximum existing numeric value (NaNs are ignored)
max_id_numeric = df_cleaned['id_numeric'].max()

# 3. Initialize the counter for new IDs, starting from the next integer.
current_new_id = int(max_id_numeric) + 1

print(f"Max numeric id_album found: {int(max_id_numeric)}. New IDs for album will start from ALB{current_new_id}")

df_cleaned.drop(columns=['id_numeric'], inplace=True)
```

```python
def normalize_id_album(df, start_new_id=999843):
    df = df.copy()

    # 1) Mark valid / invalid album IDs
    mask_valid = df["id_album"].str.match(r"^ALB\d+$", na=False)
    df.loc[~mask_valid, "id_album"] = pd.NA

    # 2) Prepare ID counter for new album IDs
    next_id = start_new_id

    # 3) Group by album identity
    groups = df.groupby(["id_artist", "album", "album_release_date"])

    for (artist, album, date), g in groups:
        rows = g.index
        vals = df.loc[rows, "id_album"].dropna()

        if len(vals) == 1:
            # Exactly one valid ID → propagate it
            df.loc[rows, "id_album"] = vals.iloc[0]

        elif len(vals) > 1:
            # Multiple IDs → choose the most frequent
            best = vals.mode().iloc[0]
            df.loc[rows, "id_album"] = best

        else:
            # No valid ID → create new one
            new_id = f"ALB{next_id}"
            next_id += 1

            df.loc[rows, "id_album"] = new_id

    print(f"✔ New album IDs assigned starting from ALB{start_new_id}")
    print(f"✔ Last used ID: ALB{next_id - 1}")

    return df
```

```python
def fill_missing_id_album(df, start_new_id=999856):
    df = df.copy()

    # Normalize incorrect album IDs (keep valid only)
    valid_mask = df["id_album"].str.match(r"^ALB\d+$", na=False)
    df.loc[~valid_mask, "id_album"] = pd.NA

    # Generate new IDs for all NA id_album
    missing_rows = df["id_album"].isna()

    count_missing = missing_rows.sum()
    print(f"→ Need to generate {count_missing} new id_album")

    # Generate new IDs
    new_ids = [f"ALB{start_new_id + i}" for i in range(count_missing)]

    # Assign in order of appearance
    df.loc[missing_rows, "id_album"] = new_ids

    print(f"✔ Assigned IDs from ALB{start_new_id} to ALB{start_new_id + count_missing - 1}")

    return df
```

```python
def validate_dates(df):
    out = df.copy()

    # --- Parse album_release_date ---
    out["parsed_date"] = pd.to_datetime(out["album_release_date"], errors="coerce")

    # --- Check if day/month/year are valid numbers ---
    out["valid_day"] = out["day"].between(1, 31, inclusive="both")
    out["valid_month"] = out["month"].between(1, 12, inclusive="both")
    out["valid_year"] = out["year"].apply(lambda x: isinstance(x, (int, float)) and x > 0)

    # Combine into one: does (Y,M,D) form a valid date?
    out["valid_ymd"] = pd.to_datetime(
        out[["year", "month", "day"]],
        errors="coerce"
    ).notna()

    # --- Compare Y/M/D with album_release_date ---
    out["match_year"] = out["parsed_date"].dt.year == out["year"]
    out["match_month"] = out["parsed_date"].dt.month == out["month"]
    out["match_day"] = out["parsed_date"].dt.day == out["day"]

    # Combine into "match all"
    out["match_all"] = out["match_year"] & out["match_month"] & out["match_day"]

    return out

df_checked = validate_dates(df_cleaned)
```

```python
def fill_ymd_from_album_date(df):
    df = df.copy()

    # chuẩn hóa về string
    ard = df["album_release_date"].astype(str).str.strip()

    # nhận diện year-only (YYYY)
    is_year_only = ard.str.fullmatch(r"\d{4}")

    # nhận diện full date (YYYY-MM-DD hoặc YYYY/MM/DD)
    is_full_date = ard.str.contains(r"-|/", regex=True)

    # --- CASE 1: album_release_date = YYYY ---
    df.loc[
        is_year_only & df["year"].isna(),
        "year"
    ] = ard[is_year_only].astype(int)

    # --- CASE 2: album_release_date = full date ---
    parsed = pd.to_datetime(
        ard.where(is_full_date),
        errors="coerce"
    )

    # fill year nếu missing
    df.loc[
        df["year"].isna() & parsed.notna(),
        "year"
    ] = parsed.dt.year

    # chỉ fill month/day khi year match
    year_match = df["year"] == parsed.dt.year

    df.loc[
        df["month"].isna() & year_match,
        "month"
    ] = parsed.dt.month

    df.loc[
        df["day"].isna() & year_match,
        "day"
    ] = parsed.dt.day

    return df
```
